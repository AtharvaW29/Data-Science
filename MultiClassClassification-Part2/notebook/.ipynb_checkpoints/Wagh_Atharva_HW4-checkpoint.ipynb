```python
extract_df = []
features = ["avg_rss12", "var_rss12", "avg_rss13", "var_rss13", "avg_rss23", "var_rss23"]
for k, v in combine_df.items():
    inner_arr = []
    for f in features:
        arr.append(v[f].describe())
        x = v[f].describe()
        inner_entry = {f'{f}_mean': x['mean'], f'{f}_std': x['std'],
                   f'{f}_min': x['min'], f'{f}_25%': x['25%'], f'{f}_50%': x['50%'],
                   f'{f}_75%': x['75%'], f'{f}_max': x['max']}
        inner_arr.append(inner_entry)
    
    entry = {'Instance': k}
    for inner in inner_arr:
        entry.update(inner)
    extract_df.append(entry)
```

```python
def bootstrap_ci(std_df, num_btstps=1000, ci=90):
    results = []
    std_df_cols = std_df.columns
    try:
        for c in std_df_cols:
            boot_stds = []
            vals = std_df[c].values
            n = len(std_df[c])
            for _ in range(num_btstps):
                resample = np.random.choice(vals, size=n, replace=True)
                boot_stds.append(np.std(resample, ddof=1))
            lower_per = (100 - ci) / 2
            upper_per = 100 - lower_per
            ci_lower = np.percentile(boot_stds, lower_per)
            ci_upper = np.percentile(boot_stds, upper_per)
            res = {'Feature': c,
                   'CI_Lower': ci_lower, 
                   'CI_Upper': ci_upper,
                   'std_estimate': np.std(vals, ddof=1),
                   "bootstrap_dist": boot_stds
                  }
            results.append(res)
    except Exception as e:
        print(f"\n There was an exception building the {ci}% CI: {e}")
    return results
```

```python
results = bootstrap_ci(std_df, 1000, 90)
res_df = pd.DataFrame(results)
res_df_cols = res_df.columns
res_df
```

```python
res_df.dtypes
```

```python
for c in res_df.itertuples(index=False):
    bootstrap_dist = np.array(c.bootstrap_dist, dtype=float)
    
    std_est = c.std_estimate
    ci_upper = c.CI_Upper
    ci_lower = c.CI_Lower
    
    plt.figure(figsize=(8,5))
    plt.hist(bootstrap_dist, bins=30, edgecolor='black', alpha=0.7) 
    
    plt.axvline(std_est, color='red', linestyle='--', label='Std Estimate')
    plt.axvline(ci_lower, color='green', linestyle='-', label='90% CI Lower')
    plt.axvline(ci_upper, color='green', linestyle='-', label='90% CI Upper')
    
    plt.title(f'Bootstrap Distribution of Std - {c.Feature}')
    plt.legend()
    plt.show()
```

```python
def set_color(instance):
    return 'red' if instance == 'bending' else 'blue'


def plot_bending(df, feature):
    pattern = rf".*{re.escape(feature)}|{re.escape('Instance')}.*"
    _df = df.filter(regex=pattern).copy()
    _df['color'] = np.where(_df['Instance'].str.contains(r'bending', regex = True), 'bending', 'Other')

    _df['index'] = _df.index
    
    fig, ax = plt.subplots(figsize=(8, 6))
    
    for key, grp in _df.groupby('color'):
        grp.plot(ax=ax, kind='scatter', x='index', y=feature, label=key, color=set_color(key), s=50)
    plt.title(f'{feature} Feature of Bending Vs Others')
    plt.legend()
    plt.show()
```

```python
def extract_features(df: pd.DataFrame, issplit, f):
    entry = {'Instance': f}
    if issplit:
        for col in df.columns:
            if col == "time_p1" or col == "time_p2":
                continue
                                
            desc = df[col].describe()
                            
            entry[f'{col}_std'] = desc['std'] 
            entry[f'{col}_mean'] = desc['mean']
            entry[f'{col}_max'] = desc['max']

        return entry
    else:
        for col in df.columns:
            if col == "time":
                continue
                                
            desc = df[col].describe()
                            
            entry[f'{col}_std'] = desc['std'] 
            entry[f'{col}_mean'] = desc['mean']
            entry[f'{col}_max'] = desc['max']

        return entry
```

```python
"""
Rough Logic for Q2(a)(ii)
Prepare the train_df dataframes for spltted plots
Here each df will undergo the following:
1. Split each df as df[:240] and [240:]
2. Create duplicate cols for the newly extracted split (6 old + 6 new = 12 features total)
3. Merge the new df and append it to the new_df_train array
4. Concat dfs in the new_df_train array.
5. Repeat the above steps till extraction and plots
I already have features = ["avg_rss12", "var_rss12", "avg_rss13", "var_rss13", "avg_rss23", "var_rss23"]
"""
def create_new_dfs(train_dir, issplit) -> pd.DataFrame:
    try:
        if os.path.isdir(train_dir):
            df_res = []
            for f in os.listdir(train_dir):
                if f.endswith('.csv'):
                    if issplit:
                        path = os.path.join(train_dir, f)
                        df = pd.read_csv(path, comment='#', names=columns, engine='python')
                        
                        n = len(df) // 2
                        
                        df_upper =df.iloc[:n].copy()
                        df_lower = df.iloc[n:].copy()
                        
                        df_upper.columns = [c + "_p1" for c in df_upper.columns]
                        df_lower.columns = [c + "_p2" for c in df_lower.columns]
                        
                        df_new_split = pd.concat([df_upper.reset_index(drop=True),
                                            df_lower.reset_index(drop=True)], axis=1)
                        
                        entry = extract_features(df_new_split, True, f)
                        
                        df_res.append(entry)
                    else:
                        path = os.path.join(train_dir, f)
                        df = pd.read_csv(path, comment='#', names=columns, engine='python')
                        entry = extract_features(df, False, f)
                        
                        df_res.append(entry)
                    
                    
            new_extract_df = pd.DataFrame(df_res)
            print('Split Applied Successfully')
            return new_extract_df
        else:
            print("Directory Not found!")
    except Exception as err:
        print(f'Error Splitting and concatenating the dfs: {err}')
```

#### iii. Time Series Segments

```python
def build_features(train_df, l, build_all=False):
    if not l: return print("l value is required.")
    if not train_df: return print("DF dict is required.")
    feature_rows = []
    if build_all==False:
        for k, v in train_df.items():
            row = {"Instance": k}
            label = 1 if "bending" in k else 0
            row["label"] = label
            for col in features:
                signals = np.array_split(v[col], l)
                    
                for idx, sig in enumerate(signals):
                    row[f"{col}_sig{idx+1}_mean"] = sig.mean()
                    row[f"{col}_sig{idx+1}_std"] = sig.std()
                    row[f"{col}_sig{idx+1}_max"] = sig.max()
                
            feature_rows.append(row)
        fdf = pd.DataFrame(feature_rows)
        X = fdf.drop(columns=["Instance", "label"])
        y = fdf["label"]
        return X, y
    else:
        activities = []
        for k, v, in train_df.items():
            row = {"Instance": k}
            arr = k.split("dataset")
            activity = arr[0]
            if activity == "bending1" or activity == "bending2":
                activity = "bending"
            row["label"] = activity
            if activity not in activities:
                activities.append(activity)
            for col in features:
                signals = np.array_split(v[col], l)
                    
                for idx, sig in enumerate(signals):
                    row[f"{col}_sig{idx+1}_mean"] = sig.mean()
                    row[f"{col}_sig{idx+1}_std"] = sig.std()
                    row[f"{col}_sig{idx+1}_max"] = sig.max()
            feature_rows.append(row)
        fdf = pd.DataFrame(feature_rows)
        X = fdf.drop(columns=["Instance", "label"])
        y = fdf["label"]
        return X, y, activities
```

```python
"""
Rough logic for Q2.(a)(iii)
1. For each df from train_df dict split at lenght of 'l'
2. Where l = [1, 2, 3, .., 20], and extract features from each signal
3. Use logistic Regression
4. Use Cross Validation
5. Choose the best number of segments
6. Select the important features properly
"""
def timeseriesmodel(train_df):
    results = {}
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    
    for l in range(1, 21):
        X, y = build_features(train_df, l)
        # print(f'Time Series Features: {X}')
        # print(f'Labels: {y}')
        n_features = X.shape[1]
        for p in range(5, min(n_features, 50), 5):
            pipeline = Pipeline([
                ("scaler", StandardScaler()),
                ("rfe", RFE(
                    estimator=LogisticRegression(max_iter=1000),
                    n_features_to_select=p
                )),
                ("model", LogisticRegression(max_iter=1000)),
            ])
            scores = cross_val_score(pipeline, X, y, cv=cv, scoring="accuracy")
            results[(l, p)] = scores.mean()
    return results
```

```python
results = timeseriesmodel(train_df)

best_pair = max(results, key=results.get)
best_score = results[best_pair]

print("Best (l, p):", best_pair)
print("Best CV Accuracy:", best_score)
```

```python
best_l = best_pair[0]
bestp = best_pair[1]
```

```python
i = 0
for k, v in results.items():
    if i < 5:
        print(f'Key: {k} -> Value: {v}')
        i += 1
    else:
        break
```

#### iv. Confusion Matrix

```python
"""
Q2(a)(iv) Rough logic
1. Rebuild the features using the best l
2. Fit pipeline using the best p and entire training data
3. Report Confusion Matrix, ROC, and AUC
4. Finally the parameters of the log regression βi and their p-values
"""
X_train_full, y_train_full = build_features(train_df, best_l)

pipeline = Pipeline([
                ("scaler", StandardScaler()),
                ("rfe", RFE(
                    estimator=LogisticRegression(max_iter=1000),
                    n_features_to_select=bestp
                )),
                ("model", LogisticRegression(max_iter=1000)),
            ])
pipeline.fit(X_train_full, y_train_full)
train_preds = pipeline.predict(X_train_full)
cm = confusion_matrix(y_train_full, train_preds)
print("Training Confusion Matrix Raw Counts: ")
print(cm)
```

```python
disp_cm = ConfusionMatrixDisplay(confusion_matrix = cm, display_labels = ['Others', 'Bending'])
disp_cm.plot(cmap=plt.cm.Blues)
plt.title('Training Confusion Matrix')
plt.show()
```

```python
"""
ROC AUC Curves for the problem
"""
probs = pipeline.predict_proba(X_train_full)[:, 1]
fpr, tpr, thresh = roc_curve(y_train_full, probs)
roc_auc = auc(fpr, tpr)
plt.figure()
plt.plot(fpr, tpr, label=f'Roc Curve (AUC={roc_auc:.2f})')
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend(loc="lower right")
plt.title('ROC Curve')
plt.show()
```

```python
X_selected = pipeline.named_steps["rfe"].transform(X_train_full)
X_selected = sm.add_constant(X_selected)
model = sm.Logit(y_train_full, X_selected).fit()
print("\n Logistic regression's βi’s and the p-values")
print(model.summary())
```

### (b) Binary Classification Using L1-penalized logistic regression

#### i. Time Series Segments

```python
"""
Q2(b)(i) Rough logic
1. Loop over l
2. Build Features
3. Loop over C (Inverse of Lambda)
4. Use Stratified 5 fold CV, and EValuate Each(l, c)
5. Rebuild the fetatures using the best l and c
6. Fit L1 training pipeline using the best c and entire training data
7. Report Confusion Matrix, ROC, and AUC
8. Finally compare with the RFE approach
"""
def l1_timeseries_model(train_df):
    results = {}
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    c_vals = [0.001, 0.01, 0.1, 1, 10, 100, 1000]
    for l in range(1, 21):
        X, y = build_features(train_df, l)
        for c in c_vals:
            pipeline = Pipeline([
                ("scaler", StandardScaler()),
                ("model", LogisticRegression(
                    penalty="l1",
                    solver="liblinear",
                    C = c,
                    max_iter=1000
                ))
            ])
            scores = cross_val_score(pipeline, X, y, cv=cv, scoring="accuracy")
            results[(l, c)] = scores.mean()
        
        return results
```

```python
l1_results = l1_timeseries_model(train_df)

best_pair_l1 = max(l1_results, key=l1_results.get)
best_score_l1 = results[best_pair_l1]
c_best = best_pair_l1[1]
l_best = best_pair_l1[0]
print("Best (l, C):", best_pair_l1)
print("Best CV Accuracy:", best_score_l1)
print("Best C: ", c_best)
print("Best L: ", l_best)
```

```python
X_train_full_l1, y_train_full_l1 = build_features(train_df, l_best)
l1_pipeline = Pipeline([
                ("scaler", StandardScaler()),
                ("model", LogisticRegression(
                    penalty="l1",
                    solver="liblinear",
                    C = c_best,
                    max_iter=1000
                ))
            ])
l1_pipeline.fit(X_train_full_l1, y_train_full_l1)
l1_train_preds = l1_pipeline.predict(X_train_full_l1)
l1_cm = confusion_matrix(y_train_full_l1, l1_train_preds)
print("Confusion Matrix for L1 approach Pipeline")
print(l1_cm)
disp_l1_cm = ConfusionMatrixDisplay(confusion_matrix = l1_cm, display_labels = ['Others', 'Bending'])
disp_l1_cm.plot(cmap=plt.cm.Blues)
plt.title('L1 Approach Confusion Matrix')
plt.show()
```

```python
"""
ROC AUC Curves for the L1 Approach
"""
l1_train_probs = l1_pipeline.predict_proba(X_train_full_l1)[:, 1]
fpr_l1, tpr_l1, thresh_l1 = roc_curve(y_train_full_l1, l1_train_probs)
roc_auc_l1 = auc(fpr_l1, tpr_l1)
plt.figure()
plt.plot(fpr_l1, tpr_l1, label=f'Roc Curve (AUC={roc_auc_l1:.2f})')
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend(loc="lower right")
plt.title('L1 Approach Training ROC Curve')
plt.show()
```

```python
X_test_l1, y_test_l1 = build_features(test_df, l_best)
l1_test_preds = l1_pipeline.predict(X_test_l1)
l1_test_cm = confusion_matrix(y_test_l1, l1_test_preds)
print("Test Confusion Matrix for L1 Approach Pipeline")
print(l1_test_cm)
disp_l1_test_cm = ConfusionMatrixDisplay(confusion_matrix = l1_test_cm, display_labels = ['Others', 'Bending'])
disp_l1_test_cm.plot(cmap=plt.cm.Blues)
plt.title('L1 Approach Confusion Matrix')
plt.show()
```

### (c) Multi-class Classification (The Realistic Case)

#### i. Time Series Segments

```python
"""
Rough logic for Q2.(c)(i)
1. Modify the Feature Builder Function to build all labels
2. Modify the Logistic Regression Function to use 'saga' solver for multinomial + L1
3. Rest Structure stays the same
"""
def multiclass_model(train_df):

    results = {}
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

    for l in range(1, 21):

        X, y = build_features(train_df, l, build_all=True)

        pipeline = Pipeline([
            ("scaler", StandardScaler()),
            ("model", LogisticRegression(
                penalty="l1",
                solver="saga",
                max_iter=5000
            ))
        ])

        scores = cross_val_score(pipeline, X, y, cv=cv, scoring="accuracy")

        results[l] = scores.mean()

    return results
```

```python
multi_class_results = multiclass_model(train_df)

best_l_multi = max(multi_class_results)

print("Best l:", best_l_multi)
```

```python
multi_X, multi_y, activities = build_features(train_df, best_l_multi, build_all=True)
multi_class_pipeline = Pipeline([
            ("scaler", StandardScaler()),
            ("model", LogisticRegression(
                penalty="l1",
                solver="saga",
                max_iter=5000
            ))
        ])
multi_class_pipeline.fit(multi_X, multi_y)
multi_class_preds = multi_class_pipeline.predict(multi_X)
multi_cm = confusion_matrix(multi_y, multi_class_preds, labels = activities)
print(multi_cm)
multi_cm_disp = ConfusionMatrixDisplay(confusion_matrix=multi_cm, display_labels=activities)
multi_cm_disp.plot(cmap="Blues")
plt.title("Confusion Matrix for Multiclass Classification [Training]")
plt.show()
```

```python
print("Precision, Recall & Accuracy score for Multi Class classification using Logistic Regression [Train data]:")
print(f"\n{classification_report(multi_y, multi_class_preds, target_names=activities)}")
```

```python
"""ROC AUC Curves for multi class classification [Training]"""
def multi_roc_auc_plot(activities, multi_y, multi_X, title=""):
    c_len = len(activities)
    y_multi_bin = label_binarize(multi_y, classes=activities)
    multi_class_probs = multi_class_pipeline.predict_proba(multi_X)
    multi_fpr, multi_tpr, multi_roc_auc = dict(), dict(), dict()
    for i in range(c_len):
        multi_fpr[i], multi_tpr[i], _ = roc_curve(y_multi_bin[:, i], multi_class_probs[:, i])
        multi_roc_auc[i] = auc(multi_fpr[i], multi_tpr[i])
    
    plt.figure()
    for i in range(c_len):
            plt.plot(multi_fpr[i], multi_tpr[i], label=f'Class {activities[i]} (AUC = {multi_roc_auc[i]:.2f})')
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(title)
    plt.legend()
    plt.show()
```

```python
multi_roc_auc_plot(activities, multi_y, multi_X, title="Multiclass ROC Curve[Training]")
```

## 3. ISLR 4.8.3

<h5>
Consider the Quadratic Discriminant Analysis (QDA) model in the
special case where there is only one feature (p = 1). Suppose there are
K classes, and if an observation belongs to class k, then:
X ~ N(μ_k, σ_k²)<br/>
To prove that the Bayes classifier in this case is not
linear, and is in fact quadratic.
</h5>
<h5>Class-Conditional Density For class k, the normal density is:</h5>
<h3>f_k(x) = (1 / sqrt(2πσ_k²)) * exp(-(x - μ_k)² / (2σ_k²))</h3>

<h5>
The Bayes classifier assigns x to the class that maximizes:
<h3>P(Y = k | X = x)</h3>
Using Bayes` theorem:
<h3>P(Y = k | X = x) ∝ π_k f_k(x)</h3>
where: <br/>
    π_k is the prior probability of class k<br/> 
    f_k(x) is the class-conditional density<br/>
Thus, we choose the class that maximizes:<br/>
<strong>π_k f_k(x)</strong></h5><br/>

<h5>Taking the log, since it is monotonic, thus maximize:<strong>log(π_k f_k(x))</strong></h5>
<div><h5><b>Substituting the density: = log π_k - (1/2) log(2πσ_k²) - (x - μ_k)² / (2σ_k²)</b></h5></div>

<h5>THerefore,</h5>
<div>(x - μ_k)² = x² - 2xμ_k + μ_k²</div>
<h5>Substituting,</h5>
<div> -(1 / (2σ_k²)) (x² - 2xμ_k + μ_k²) </div>
<div>= -x² / (2σ_k²) + xμ_k/σ_k² - μ_k² / (2σ_k²)</div>


<h5>Putting everything together:</h5>
<div>δ_k(x) = - x² / (2σ_k²) + xμ_k / σ_k² - μ_k² / (2σ_k²) - (1/2) logσ_k² + log π_k</div>


<h5><strong>
    Because the discriminant function contains a class-dependent quadratic<br/>
    term in x when variances differ, the Bayes classifier in QDA is quadratic<br/> 
    rather than linear. Thus, in the one-dimensional case with class-specific<br/> 
    variances, the Bayes decision boundary is quadratic.
</strong></h5>


## 4. ISLR 4.8.7

<div>
    <h5><strong> To Find P(Yes| X = 4)</strong></h5>
    <ul> Where;
        <li>μYes = 10</li>
        <li>μNo = 0</li>
        <li>σ**2 = 36 ⇒ σ = 6</li>
        <li>πYes = 0.8</li>
        <li>πNo=0.2</li>
    </ul
</div>

<div>
    <h5> As Per Baye's Theorem<h5>
        <div>
            P(Yes∣X=4) =πYesfYes(4) / (πNofNo(4) + πYesfYes(4))
        </div>
</div>

<div>
    <h5>As per Normal Density Formula </h5>
    <div>
        f(x)= 1 / (2πσ2)^0.5 e**(−(x−μ)**2 / 2σ2)
    </div>
    <h5>Since both classes have the same variance, the constant term cancels in the ratio.</h5>
</div>

<div>
    <h5>Computing Exponents For Yes (μ = 10)</h5>
    <div>
        (4−10)**2 = (−6)**2 = 36
    </div>
    <div>
        36 / (2 * 36) = 36/72 = 0.5
    </div>
    <h5> So, fYes(4)  ∝ e**−0.5 </h5>
    <h5> For No (μ = 0)</h5>
    <div>
        (4−0)**2 = 16
    </div>
    <div>
        16/72 = 0.222
    </div>
    <h5> So, fNo(4)  ∝ e**−0.222 </h5>
    <div><h5>e**−0.5 ≈ 0.607</h5></div>
    <div><h5>e**−0.2222 ≈ 0.801</h5></div>
</div>

<div>
    <h5> Plugging the values into the Naeve bayes Formula </h5>
    <div>
        P(Yes | X = 4) = 0.8 x 0.607/(0.8 x 0.607 + 0.2 x 0.801)
        P(Yes | X = 4) = 0.752 ~ 0.75
    </div>
    <h5><strong> There is a 75% probability that the company will issue a dividend </strong></h5>
</div>
