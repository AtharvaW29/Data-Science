#### Q1 (c)

```python
def tree_representation(tree, feature_names, target_names=None, class_names=("no", "yes")):
    """
    We will have to recursively traverse the tree and specify each condition/decision while doing so
    I will use dfs with a recursive approach
    """
    
    tree_ = tree.tree_
    feature_name = [
        feature_names[i] if i != _tree.TREE_UNDEFINED else "undefined"
        for i in tree_.feature
    ]
    print("def tree({}):".format(", ".join(feature_names)))
    
    def recurse(node, depth):
        indent = " " * depth
        if tree_.feature[node] != _tree.TREE_UNDEFINED:
            name = feature_name[node]
            threshold = tree_.threshold[node]
            print(f"{indent}if {name} <= {threshold:.3f}:")
            recurse(tree_.children_left[node], depth + 1)
            print(f"{indent}else:  # if {name} > {threshold:.3f}")
            recurse(tree_.children_right[node], depth + 1)
        else:
            counts = np.asarray(tree_.value[node])
            if counts.ndim == 1:
                counts = counts[np.newaxis, :]
            preds = counts.argmax(axis=1)

            if target_names is None:
                print(f"{indent}return {preds.tolist()}  # counts={counts.tolist()}")
            else:
                label_text = ", ".join(
                    f"{t}={class_names[p]}" for t, p in zip(target_names, preds)
                )
                print(f"{indent}return {label_text}  # counts={counts.tolist()}")

    recurse(0, 1)
```

```python
tree_representation(dt, features, target)
```

#### Q1. (d)

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
base_tree = DecisionTreeClassifier(random_state=42)
pruning_path = base_tree.cost_complexity_pruning_path(X_train, y_train)
```

```python
ccp_alphas = pruning_path.ccp_alphas
impurities = pruning_path.impurities

ccp_alphas = ccp_alphas[:-1]
```

```python
clfs = []
train_scores = []
test_scores = []
node_counts = []
depths = []
```

```python
for alpha in ccp_alphas:
    clf = DecisionTreeClassifier(random_state=42, ccp_alpha=alpha)
    clf.fit(X_train, y_train)
    clfs.append(clf)
    train_scores.append(clf.score(X_train, y_train))
    test_scores.append(clf.score(X_test, y_test))
    node_counts.append(clf.tree_.node_count)
    depths.append(clf.tree_.max_depth)
```

```python
best_idx = int(np.argmax(test_scores))
best_alpha = ccp_alphas[best_idx]
best_tree = clfs[best_idx]

print("Chosen ccp_alpha:", best_alpha)
print("Node count:", best_tree.tree_.node_count)
print("Max depth:", best_tree.tree_.max_depth)
print("Train score:", train_scores[best_idx])
print("Test score:", test_scores[best_idx])
```

```python
plt.figure(figsize=(7, 4))
plt.plot(ccp_alphas, train_scores, marker='o', label='train')
plt.plot(ccp_alphas, test_scores, marker='o', label='test')
plt.xlabel("ccp_alpha")
plt.ylabel("score")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16, 10))
plot_tree(
    best_tree,
    feature_names=features,
    filled=True,
    rounded=True,
    impurity=True
)
plt.tight_layout()
plt.show()
```

```python
tree_representation(best_tree, features, target)
```

#### Q2.(f)

```python
clean_df.iloc[:, -1].shape
```

```python
simple_lm = LinearRegression()
X_2 = clean_df.loc[:, top_11_cols]
X_train_2, X_test_2, y_train_2, y_test_2 = train_test_split(
    X_2, clean_df.iloc[:, -1], test_size=0.2, random_state=42
)
simple_lm.fit(X_train_2, y_train_2)
```

```python
simple_lm_pred = simple_lm.predict(X_test_2)
def print_metrics(_pred, _test)->None:
    mse_ = mean_squared_error(_test, _pred)
    rmse_ = np.sqrt(mse_)
    r2_ = r2_score(_test, _pred)
    
    print(f"Test Mean Squared Error: {mse_}")
    print(f"Test Root Mean Squared Error: {rmse_}")
    print(f"Test R-squared Score: {r2_}")
```

```python
print_metrics(simple_lm_pred, y_test_2)
```

#### Q2.(h)

##### LASSO without Standardization

```python
lasso_model = LassoCV(cv=5, random_state=42)
lasso_model.fit(X_train_2, y_train_2)
print(f"Best alpha (no scaling): {lasso_model.alpha_}")
y_lasso_pred = lasso_model.predict(X_test_2)
```

```python
"""For LASSO Without Standardization WE have the Test Error as"""
print_metrics(y_lasso_pred, y_test_2)
```

```python
coef_ = lasso_model.coef_
selected_features_ = X_2.columns[coef_ != 0]

print("Selected features:")
print(selected_features_.tolist())
```

#### LASSO with Standardization

```python
lasso_scaled = Pipeline([
    ('scaler', StandardScaler()),
    ('lasso', LassoCV(cv=5, random_state=42))
])
lasso_scaled.fit(X_train_2, y_train_2)
best_alpha_scaled = lasso_scaled.named_steps['lasso'].alpha_
print(f"Best Alpha (Scaled): {best_alpha_scaled}")
y_lasso_pred_scaled = lasso_scaled.predict(X_test_2)
```

```python
"""For LASSO with standadrdization (with scaling), we have the following test error"""
print_metrics(y_lasso_pred_scaled, y_test_2)
```

```python
coef_scaled = lasso_scaled.named_steps['lasso'].coef_
selected_features_2 = X_2.columns[coef_scaled != 0]

print("Selected features (scaled):")
print(selected_features_2.tolist())
```

<ul>
<li>A LASSO model was trained with the regularization parameter λ selected via cross-validation.</li> 
<li>Without feature standardization, the model achieved a test MSE of 0.03115 and selected 10 features. 
<li>After standardizing the features, the model achieved a test MSE of 0.03114 and selected 11 features.</li> 
<li>The standardized model performed better because LASSO is sensitive to feature scaling, and standardization     ensures fair penalization across all features.</li> 
<li>Additionally, the standardized model provided a more reliable and interpretable subset of features.</li>
</ul>

#### Q2.(i)

```python
pcr_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('pca', PCA()),
    ('reg', LinearRegression())
])

param_grid = {
    "pca__n_components": list(range(1, X_train_2.shape[1] + 1))
}

grid_ = GridSearchCV(
    pcr_pipeline,
    param_grid = param_grid,
    cv=5,
    scoring='neg_mean_squared_error',
)

grid_.fit(X_train_2, y_train_2)
best_m = grid_.best_params_["pca__n_components"]
print(f"Best Number of Principal Components: {best_m}")
y_pcr_pred = grid_.predict(X_test_2)
```

```python
print("Test MSE for PCR model")
print_metrics(y_pcr_pred, y_test_2)
```

#### Q2(j)

##### Choosing XGBoost, because it is a regularized boosting tree algorithm that incorporates 
##### L1 regularization (as well as L2 regularization) into its objective function to control model complexity
##### and prevent overfitting

```python
xgb = XGBRegressor(
    objective="reg:squarederror",
    tree_method="hist",
    random_state=42,
    n_estimators=300,
    learning_rate=0.05,
    max_depth=4,
    subsample=0.8,
    colsample_bytree=0.8,
    n_jobs=-1
)

param_ = {
    "reg_alpha": np.logspace(-4, 2, 20)
}

cross_val = KFold(n_splits=5, shuffle=True, random_state=42)

xgb_grid = GridSearchCV(
    estimator=xgb,
    param_grid=param_,
    scoring="neg_mean_squared_error",
    cv=cross_val,
    n_jobs=-1
)

xgb_grid.fit(X_train_2, y_train_2)

best_xgb_model=xgb_grid.best_estimator_
y_pred_best_xgb = best_xgb_model.predict(X_test_2)
print("Best reg_alpha:", xgb_grid.best_params_["reg_alpha"])
print("XGB Boosting Tree Test MSE:")
print_metrics(y_pred_best_xgb, y_test_2)
```
